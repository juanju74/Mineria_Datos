# Exploratory Data Analysis inspirado en DataExplorer

Este notebook explica cómo usar las ideas del proyecto **DataExplorer** (overview, tipos de datos, valores faltantes, distribuciones, correlaciones y detección básica de outliers) usando Python en Jupyter.

Dataset usado: `educacion_estadisticas.csv` del archivo ZIP suministrado.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize']=(10,5)
df = pd.read_csv('data/educacion_estadisticas.csv')
df.head()
```

## 1. Estructura general (similar a `introduce()` en DataExplorer)

```python
print('Filas y columnas:', df.shape)
print('\nTipos de datos:')
print(df.dtypes.value_counts())
print('\nPrimeras columnas:')
df.dtypes.head(10)
```

## 2. Resumen estadístico

```python
df.describe().T.head(15)
```

## 3. Valores faltantes (similar a `plot_missing()`)

```python
missing = df.isna().sum().sort_values(ascending=False)
missing = missing[missing>0]
missing
```

```python
if len(missing)>0:
    missing.plot(kind='bar')
    plt.title('Valores faltantes por columna')
    plt.show()
else:
    print('No se detectaron valores faltantes.')
```

## 4. Distribución de una variable numérica (similar a `plot_histogram()`)

```python
df['desercion'].hist(bins=20)
plt.title('Distribución de la deserción')
plt.xlabel('Deserción')
plt.show()
```

## 5. Comparación por categoría

```python
top = df.groupby('departamento')['desercion'].mean().sort_values(ascending=False).head(10)
top.plot(kind='bar')
plt.title('Top 10 departamentos con mayor deserción promedio')
plt.ylabel('Deserción promedio')
plt.show()
```

## 6. Correlación (similar a `plot_correlation()`)

```python
corr = df.select_dtypes(include='number').corr(numeric_only=True)
focus = corr[['desercion']].sort_values('desercion', ascending=False).head(10)
focus
```

```python
plt.imshow(corr.iloc[:10,:10], aspect='auto')
plt.title('Mapa simple de correlación (primeras 10 vars)')
plt.colorbar()
plt.show()
```

## 7. Detección básica de outliers

```python
q1 = df['desercion'].quantile(0.25)
q3 = df['desercion'].quantile(0.75)
iqr = q3-q1
outliers = df[(df['desercion'] < q1-1.5*iqr) | (df['desercion'] > q3+1.5*iqr)]
outliers[['departamento','ano','desercion']].head(10)
```
